## Q1 — Write and Read Back

> Write `write_and_read(data, filename)` that:
> 1. Saves list of dicts to `filename.csv`
> 2. Saves same data to `filename.json`
> 3. Reads both back
> 4. Prints row count: Original / CSV back / JSON back

> Test with 3 employee records.


````python
import csv, json

# YOUR CODE:


# Test data
employees = [
    {'id':1, 'name':'Raj',   'salary':85000},
    {'id':2, 'name':'Priya', 'salary':72000},
    {'id':3, 'name':'Amit',  'salary':92000},
]

def write_and_read(data, filename):
    with open(f"{filename}.csv","w",newline="")as f:
        a=csv.DictWriter(f,fieldnames=data[0].keys())
        a.writeheader()
        a.writerows(data)
    with open(f"{filename}.csv","r") as f:
        a=list(csv.DictReader(f))

    with open(f"{filename}.json","w")as j:
        b=json.dump(data,j)

    with open(f"{filename}.json","r")as j:
        c=json.load(j)
    print(len(data),len(a),len(c))



write_and_read(employees,'employees')
    

# Expected output:
# Original: 3
# CSV back: 3
# JSON back: 3
````

## Q2 — Safe Reader

> Write `read_csv_safe(filepath)` that:
> - Returns list of dicts if file exists and has data
> - Prints 'File not found' and returns [] if file missing
> - Prints 'File is empty' and returns [] if file has no rows
> - Never crashes

> Test 3 cases: valid file, fake path, empty file.


````python
import csv

# YOUR CODE:

def read_csv_safe(filepath):
    try:
      with open(filepath,"r") as f:
            c=list(csv.DictReader(f))
            if len(c)==0:
                print("file is empty")
                return []
            return c
    except FileNotFoundError:
        print("file not found")
        return[]

# Test 1: valid file (use employees.csv from Q1)
print(read_csv_safe('employees.csv'))

# Test 2: file doesn't exist
print(read_csv_safe('fake_file.csv'))

# Test 3: empty file
with open('empty.csv', 'w') as f:
    pass  # creates empty file
print(read_csv_safe('empty.csv'))
````

## Q3 — Filter and Save

> You have orders data. Do these steps:
> 1. Save raw orders to `raw_orders.csv`
> 2. Read it back
> 3. Filter only `completed` orders
> 4. Save filtered data to `completed_orders.csv`
> 5. Print: total orders / completed orders / saved to file


````python
# YOUR CODE:
# Step 1: save orders to raw_orders.csv
# Step 2: read it back
# Step 3: filter completed only
# Step 4: save filtered to completed_orders.csv
# Step 5: print counts

# Expected output:
# Total orders: 6
# Completed orders: 3
# Saved to completed_orders.csv


import csv

orders = [
    {'id':1, 'city':'Mumbai',    'amount':5000,  'status':'completed'},
    {'id':2, 'city':'Delhi',     'amount':3000,  'status':'pending'},
    {'id':3, 'city':'Mumbai',    'amount':8000,  'status':'completed'},
    {'id':4, 'city':'Bangalore', 'amount':2000,  'status':'cancelled'},
    {'id':5, 'city':'Delhi',     'amount':12000, 'status':'completed'},
    {'id':6, 'city':'Pune',      'amount':4500,  'status':'pending'},
]

with open("raw_order.csv","w",newline="") as f:
    c=csv.DictWriter(f,fieldnames=orders[0].keys())
    c.writeheader()
    c.writerows(orders)

with open("raw_order.csv","r") as f:
    z=list(csv.DictReader(f))
    print(z)

with open("completed_order.csv","w",newline="") as f:
    c=csv.DictWriter(f,fieldnames=z[0].keys())
    c.writeheader()
    for i in z:
        if i["status"]== "completed":
            c.writerow(i)

with open("completed_order.csv","r") as f:
    q=list(csv.DictReader(f))
    print(q)







````
