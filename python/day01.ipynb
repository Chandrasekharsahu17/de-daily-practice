```python
# ─────────────────────────────────────────
# PYTHON Q1
# Given a list of 8 salaries, write a function that returns only salaries above a given threshold. Call it twice with different thresholds.
# ─────────────────────────────────────────

# YOUR CODE:
def mouuu(a,z):
    result=[]
    for i in a:
        if i>z:
            result.append(i)
    return result


salaries = [133333, 14156456, 15641, 1, 156, 1651056, 56316, 165165]
# Call 1
print("Above 10000:")
print(mouuu(salaries, 10000))

# Call 2
print("\nAbove 100000:")
print(mouuu(salaries, 100000))
```

```python
# ─────────────────────────────────────────
# PYTHON Q2
# Write a function clean_names(names) that strips whitespace, converts to title case, and removes duplicates. Test with messy data.
# ─────────────────────────────────────────

# YOUR CODE:

def clean_names(names):
    names=names.strip().title()
    
    z=""
    for i in names:
        if i not in z:
            z+=i
    return z

           
    return names,z


print(clean_names("  Aanna  "))


def clean_names_list(names):
    clean=[]
    for n in names:
        n=n.strip().title()
        if n not in clean:
           clean.append(n)
    return clean


names = [
    "  john",
    "JOHN ",
    " alice ",
    "ALICE",
    " bob",
    "Bob "
]

clean_names_list(names)   
```
