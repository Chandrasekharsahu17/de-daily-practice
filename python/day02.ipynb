```python
# ─────────────────────────────────────────
# PYTHON Q1
# Given a list of order dicts, write a function group_by_status(orders) that groups them into a dict like {'completed': [...], 'pending': [...]}.
# ─────────────────────────────────────────

# YOUR CODE:
orders = [
    {"order_id": 1, "status": "completed"},
    {"order_id": 2, "status": "pending"},
    {"order_id": 3, "status": "completed"},
    {"order_id": 4, "status": "cancelled"},
    {"order_id": 5, "status": "pending"}
]

def group_by_status(orders):
    a={}
    b=1
    for i in orders:
        if i["status"] not in a:
            a[i["status"]] =[]
        a[i["status"]].append(i["order_id"])

    return a
c=group_by_status(orders)   
print(c)
```

```python
# ─────────────────────────────────────────
# PYTHON Q2
# Given two lists of customer_ids (yesterday and today) use sets to find: new customers, lost customers, retained customers.
# ─────────────────────────────────────────

# YOUR CODE:
yesterday = [101, 102, 103, 104, 105]
today = [103, 104, 105, 106, 107]

retained_customers=set(yesterday).intersection(set(today))
new_customer=set(today).difference(set(yesterday))
lost_cutomer=set(yesterday).difference(set(today))

print(f"the r_c{retained_customers} n_c{new_customer}and l_c{lost_cutomer}")
```
