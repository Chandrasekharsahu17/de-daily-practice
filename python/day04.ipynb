```python
# ─────────────────────────────────────────
# PYTHON Q1
# Given a list of order amounts use map() to apply 18% GST and filter() to keep only amounts above 5000.
# ─────────────────────────────────────────

# YOUR CODE:
lst=[124,42,42,4,5,634,2,6,7234,41334,516,4316,463,643,65746,763,46]
def gst(x):
    return x*1.18
def fil(a):
    if a>5000:
        return a
a=list(map(gst,list(filter(fil,lst))))
print(a)
```

```python
# ─────────────────────────────────────────
# PYTHON Q2
# Use lambda with sorted() to sort a list of employee dicts: first by dept_id ascending then by salary descending.
# ─────────────────────────────────────────

# YOUR CODE:



employees = [
    {'name': 'A', 'dept_id': 2, 'salary': 50000},
    {'name': 'B', 'dept_id': 1, 'salary': 60000},
    {'name': 'C', 'dept_id': 1, 'salary': 50000},
    {'name': 'D', 'dept_id': 2, 'salary': 70000}
]
a= sorted(employees ,key=lambda employees :(employees['dept_id'],-employees['salary']))
print(a)
```
