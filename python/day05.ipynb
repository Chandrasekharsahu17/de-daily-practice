```python
# ─────────────────────────────────────────
# PYTHON Q1
# Use reduce() from functools to calculate total revenue from a list of order dicts. Then rewrite with sum() — compare both.
# ─────────────────────────────────────────

# YOUR CODE:
from functools import reduce
orders = [
    {"order_id": 101, "revenue": 5000},
    {"order_id": 102, "revenue": 3000},
    {"order_id": 103, "revenue": 7000},
    {"order_id": 104, "revenue": 2000}
]

b=reduce(lambda a,x: a+x["revenue"],orders,0)

print(b)

c=sum(i["revenue"] for i in orders )
print(c)
```

```python
# ─────────────────────────────────────────
# PYTHON Q2
# Write a pipeline using map, filter, reduce: start with raw orders list, filter valid ones, map to extract amount, reduce to total. One chain.
# ─────────────────────────────────────────

# YOUR CODE:



orders = [
    {"id": 1, "amount": 5000, "valid": True},
    {"id": 2, "amount": 3000, "valid": False},
    {"id": 3, "amount": 7000, "valid": True},
    {"id": 4, "amount": 2000, "valid": True}
]

a=reduce(map(lambda orders:orders["amount"],filter(lambda orders:orders["valid"],orders)))
print(a)
```
